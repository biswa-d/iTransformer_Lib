```python
# Check the relationship between true1 and true2
def check_subset(true1, true2):
    try:
        # Find where true2 starts in true1
        start_idx = true1.tolist().index(true2.iloc[0])
        
        # Verify sequential match
        matches = true1.iloc[start_idx:start_idx + len(true2)].equals(true2)
        if matches:
            print(f"true2 is a subset of true1 starting at index {start_idx}.")
        else:
            print("true2 does not match sequentially with true1 after its start index.")
    except ValueError:
        print("true2's first value is not found in true1.")
```

```python
from datetime import datetime, timedelta

def add_date_column(file_path, output_path, start_date):
    """
    Reads a CSV file, adds a 'date' column with 1-second increments from the start_date,
    and saves the modified DataFrame to a new CSV file.

    Parameters:
        file_path (str): Path to the input CSV file.
        output_path (str): Path to save the output CSV file.
        start_date (str): Starting date in "YYYY-MM-DD HH:MM:SS" format.
    """
    try:
        # Read the CSV file
        df = pd.read_csv(file_path)
        
        # Convert start_date to a datetime object
        start_datetime = datetime.strptime(start_date, "%Y-%m-%d %H:%M:%S")
        
        # Create a 'date' column with 1-second increments
        df['date'] = [start_datetime + timedelta(seconds=i) for i in range(len(df))]
        
        # Save the modified DataFrame to a new CSV file
        df.to_csv(output_path, index=False)
        print(f"File saved successfully at: {output_path}")
    except Exception as e:
        print(f"Error: {e}")
```
